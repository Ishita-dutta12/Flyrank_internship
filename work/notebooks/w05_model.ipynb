## 1. Method choice and why

*Which method from the toolkit, and why it fits your lane.*

Text Answer (Markdown cell above code):

Selected Random Forest Classifier because our FlyRank opportunity-scoring lane relies on tabular, heterogeneous search metrics (impressions, average position, click-through trends, content age) with non-linear interactions. Unlike linear models or rigid single-split decision trees, an ensemble of trees captures complex feature interactions (e.g., high impressions combined with dropping positions) without requiring extensive manual feature scaling or hyperparameter tuning, making it robust against overfitting on our intermediate-size sample.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import precision_score, roc_auc_score, average_precision_score

# Initialize Random Forest model with controlled depth and reproducibility seed
model = RandomForestClassifier(
    n_estimators=100,
    max_depth=6,
    random_state=42,
    class_weight="balanced"
)
print("Model initialized successfully:", type(model).__name__)
```
