### 1. Ranked Actions and Reason Codes

We translate model scores into a prioritized action queue. Each recommendation is tagged with a clear reason code to ensure explainability for content teams.

* **Top Recommendations:** High-priority items identified by the validated model.
* **Reason Codes:**
  * `RC_VISIBILITY_DROP`: Rapid decline in ranking position over the observation window.
  * `RC_HIGH_INTENT`: High search volume combined with low current content coverage.
  * `RC_DECAY_STALE`: Content freshness score below threshold, requiring updates.

```python
import pandas as pd

# Generate ranked action queue dataframe
ranked_actions = pd.DataFrame({
    "content_id": ["C101", "C102", "C103"],
    "archetype": ["Pillar Guide", "Comparison", "FAQ Update"],
    "priority_score": [0.94, 0.88, 0.81],
    "reason_code": ["RC_HIGH_INTENT", "RC_VISIBILITY_DROP", "RC_DECAY_STALE"],
    "recommended_action": ["Expand depth & add examples", "Refresh comparison table", "Update outdated stats"]
})
print(ranked_actions)
```
